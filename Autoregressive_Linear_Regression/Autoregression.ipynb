```
# Load raw ultrasonic data
data1 = np.load('/content/drive/My Drive/Colab Notebooks/p5270/Labquake_prediction/Data_Extraction/Data/NPData/Npclean1.npz')
data2 = np.load('/content/drive/My Drive/Colab Notebooks/p5270/Labquake_prediction/Data_Extraction/Data/NPData/Npclean2.npz')
data3 = np.load('/content/drive/My Drive/Colab Notebooks/p5270/Labquake_prediction/Data_Extraction/Data/NPData/Npclean3.npz')
data4 = np.load('/content/drive/My Drive/Colab Notebooks/p5270/Labquake_prediction/Data_Extraction/Data/NPData/Npclean4.npz')
data5 = np.load('/content/drive/My Drive/Colab Notebooks/p5270/Labquake_prediction/Data_Extraction/Data/NPData/Npclean5.npz')
```

```
# Get training time
def truncate(n,i):
  return math.floor(n*10**i)/10**i

train_time = np.concatenate((data1['Time'], data2['Time'], data3['Time'], data4['Time'], data5['Time']), axis=None)
ttime = [truncate(i,2) for i in train_time]
print(np.shape(ttime))
ttime = np.reshape(ttime, (np.shape(ttime)[0], 1))
print(ttime)
print(np.shape(ttime))

del train_time
```

```
# Collect raw ultrasonic data
train_data = np.concatenate((data1['AcData'], data2['AcData'], data3['AcData'], data4['AcData'], data5['AcData']), axis=0)
print(train_data)
print(np.shape(train_data))
```

```
# Delete raw ultrasonic data
del data1, data2, data3, data4, data5
```

```
# Combine ultrasound data & time
ttime = pd.DataFrame(ttime).rename(columns={0:'Time'})
full_df = ttime.join(pd.DataFrame(train_data))
print(full_df)
del train_data, ttime
```

```
# Drop duplicate time data
full_df = full_df.drop_duplicates(subset=['Time'], keep='last').reset_index(drop=True)
full_df = full_df.set_index('Time')
print(full_df)
```

```
# Load shear stress data
data = loadmat('/content/drive/My Drive/Colab Notebooks/p5270/Labquake_prediction/MLpreprocessed_code/data/p5270_ML_Master.mat')
m_t = pd.DataFrame({'Time':data['Time'].ravel().round(2), 'SS':data['SS'].ravel()})
m_t = m_t[:][175604:175604+132400].reset_index(drop=True)
print(m_t)
```

```
# Merge raw ultrasonic data & shear stress based on time
# Prepare input data (xdf): raw ultrasonic signal
# Prepare output data (ydf): shear stress
df = full_df.reset_index().merge(m_t, on='Time').reset_index(drop=True)
xdf = df.drop(['Time', 'SS'], axis=1)
ydf = df[['SS']]
print(xdf)
print(ydf)
```

```
# Split data in train-val-test (72-8-20)
x_train, x_test1, y_train, y_test = train_test_split(xdf, ydf, test_size=0.2, shuffle=False)
X_train1, X_val1, Y_train, Y_val = train_test_split(x_train, y_train, test_size=0.125, shuffle=False)

del x_train, y_train
```

```
# Reshape data & scaling
X_train = np.reshape(X_train1.values, (-1,1))
X_val = np.reshape(X_val1.values, (-1,1))
x_test = np.reshape(x_test1.values, (-1,1))

del X_train1, X_val1, x_test1

sc = MinMaxScaler(feature_range=(-1,1))
X_train = sc.fit_transform(X_train)
X_val = sc.transform(X_val)
x_test = sc.transform(x_test)

X_train = np.reshape(X_train, (-1,1568))
X_val = np.reshape(X_val, (-1,1568))
x_test = np.reshape(x_test, (-1,1568))

print(X_train.shape, X_val.shape, x_test.shape)
print(Y_train.shape, Y_val.shape,y_test.shape)
```

```
# Autoregression Models

# Average method
y_pred_avg_train = np.full(len(Y_train), np.mean(Y_train))
y_pred_avg_val = np.full(len(Y_val), np.mean(Y_train))
y_pred_avg_test = np.full(len(y_test), np.mean(Y_train))

r2_avg_train = r2_score(Y_train, y_pred_avg_train)
rmse_avg_train = np.sqrt(mean_squared_error(Y_train, y_pred_avg_train))
r2_avg_val = r2_score(Y_val, y_pred_avg_val)
rmse_avg_val = np.sqrt(mean_squared_error(Y_val, y_pred_avg_val))
r2_avg_test = r2_score(y_test, y_pred_avg_test)
rmse_avg_test = np.sqrt(mean_squared_error(y_test, y_pred_avg_test))

# Rolling average method
rolling_avg_pred_value = pd.DataFrame(Y_train).rolling(300).mean().iloc[-1]
y_pred_rolling_avg_train = np.full(len(Y_train), rolling_avg_pred_value)
y_pred_rolling_avg_val = np.full(len(Y_val), rolling_avg_pred_value)
y_pred_rolling_avg_test = np.full(len(y_test), rolling_avg_pred_value)

r2_rolling_avg_train = r2_score(Y_train, y_pred_rolling_avg_train)
rmse_rolling_avg_train = np.sqrt(mean_squared_error(Y_train, y_pred_rolling_avg_train))
r2_rolling_avg_val = r2_score(Y_val, y_pred_rolling_avg_val)
rmse_rolling_avg_val = np.sqrt(mean_squared_error(Y_val, y_pred_rolling_avg_val))
r2_rolling_avg_test = r2_score(y_test, y_pred_rolling_avg_test)
rmse_rolling_avg_test = np.sqrt(mean_squared_error(y_test, y_pred_rolling_avg_test))

# Naive forecast method
naive_pred_value = Y_train.iloc[len(Y_train)-1]
y_pred_naive_train = np.full(len(Y_train), naive_pred_value)
y_pred_naive_val = np.full(len(Y_val), naive_pred_value)
y_pred_naive_test = np.full(len(y_test), naive_pred_value)

r2_naive_train = r2_score(Y_train, y_pred_naive_train)
rmse_naive_train = np.sqrt(mean_squared_error(Y_train, y_pred_naive_train))
r2_naive_val = r2_score(Y_val, y_pred_naive_val)
rmse_naive_val = np.sqrt(mean_squared_error(Y_val, y_pred_naive_val))
r2_naive_test = r2_score(y_test, y_pred_naive_test)
rmse_naive_test = np.sqrt(mean_squared_error(y_test, y_pred_naive_test))

print("1. AVERAGE METHOD:")
print("  TRAIN: RMSE: ", rmse_avg_train, "\nR2 score: ", r2_avg_train)
print("  VALIDATION: RMSE: ", rmse_avg_val, "\nR2 score: ", r2_avg_val)
print("  TEST: RMSE: ", rmse_avg_test, "\nR2 score: ", r2_avg_test)

print("\n2. ROLLING AVERAGE METHOD:")
print("  TRAIN: RMSE: ", rmse_rolling_avg_train, "\nR2 score: ", r2_rolling_avg_train)
print("  VALIDATION: RMSE: ", rmse_rolling_avg_val, "\nR2 score: ", r2_rolling_avg_val)
print("  TEST: RMSE: ", rmse_rolling_avg_test, "\nR2 score: ", r2_rolling_avg_test)

print("\n3. NAIVE FORECAST METHOD:")
print("  TRAIN: RMSE: ", rmse_naive_train, "\nR2 score: ", r2_naive_train)
print("  VALIDATION: RMSE: ", rmse_naive_val, "\nR2 score: ", r2_naive_val)
print("  TEST: RMSE: ", rmse_naive_test, "\nR2 score: ", r2_naive_test)

'''
  # ARIMA
model2 = ARMA(Y_train, order=(2,1)).fit(disp=0)

# ARIMA prediction should be handled differently, as it's a time series model
# and would typically predict based on previous actual values or a forecast horizon.
# For simplicity, if we were to treat it like the others, we would predict from the last Y_train value
# or a more complex forecasting strategy.
# Since the ARIMA model is commented out, I will leave it as is.

# y_pred_arima = model2.predict(start=y_test.index[0], end=y_test.index[-1])
# r2_arima = r2_score(y_test, y_pred_arima)
# rmse_arima = np.sqrt(mean_squared_error(y_test, y_pred_arima))

# print("\n4. ARMA METHOD: \nRMSE: ", rmse_arima, "\nR2 score: ", r2_arima)
'''
```
