```python
# Load library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
# Function to load data
def readcsv(directory_path, file_name, column_name):
    full_path = os.path.join(directory_path, file_name)
    df = pd.read_csv(full_path)
    return df[column_name]
```

```python
# Load data: Seed 0
path_seed_0 = '/content/drive/My Drive/Colab Notebooks/Final_Features_Extraction_Codes/Uncertainty_Quantifications/Hand_Crafted_Model/Seed_0/Files'
Time_Test_Seed_0 = readcsv(path_seed_0,"testing_results.csv",'Time_s')
Actual_SS_Seed_0 = readcsv(path_seed_0,"testing_results.csv",'Actual_Shear_Stress')
Predicted_SS_Seed_0 = readcsv(path_seed_0,"testing_results.csv",'Predicted_Shear_Stress')

plt.plot(Time_Test_Seed_0,Actual_SS_Seed_0,color='black',label='Actual')
plt.plot(Time_Test_Seed_0,Predicted_SS_Seed_0,color='red',label='Predicted')
plt.xlabel('Time (s)')
plt.ylabel('Shear Stress (MPa)')
plt.legend()
plt.axis('tight')
plt.show()
```

```python
# Load data: Seed 40
path_seed_40 = '/content/drive/My Drive/Colab Notebooks/Final_Features_Extraction_Codes/Uncertainty_Quantifications/Hand_Crafted_Model/Seed_40/Files'
Time_Test_Seed_40 = readcsv(path_seed_40,"testing_results.csv",'Time_s')
Actual_SS_Seed_40 = readcsv(path_seed_40,"testing_results.csv",'Actual_Shear_Stress')
Predicted_SS_Seed_40 = readcsv(path_seed_40,"testing_results.csv",'Predicted_Shear_Stress')

plt.plot(Time_Test_Seed_40,Actual_SS_Seed_40,color='black',label='Actual')
plt.plot(Time_Test_Seed_40,Predicted_SS_Seed_40,color='red',label='Predicted')
plt.xlabel('Time (s)')
plt.ylabel('Shear Stress (MPa)')
plt.legend()
plt.axis('tight')
plt.show()
```

```python
# plt.rcParams['figure.dpi'] = 600 # Adjust DPI as needed (e.g., 100, 150, 200)

# Combine predicted shear stress from all seeds into a DataFrame
predicted_ss_df = pd.DataFrame({
    'Seed_0': Predicted_SS_Seed_0,
    'Seed_10': Predicted_SS_Seed_10,
    'Seed_20': Predicted_SS_Seed_20,
    'Seed_30': Predicted_SS_Seed_30,
    'Seed_40': Predicted_SS_Seed_40
})

# Calculate the mean and standard deviation of the predictions
mean_predicted_ss = predicted_ss_df.mean(axis=1);
std_predicted_ss = predicted_ss_df.std(axis=1);

# Plotting
# plt.figure(figsize=(8, 5))
plt.figure(figsize=(14, 5))
plt.plot(Time_Test_Seed_0, Actual_SS_Seed_0, color='black', label='Ground Truth')
plt.plot(Time_Test_Seed_0, mean_predicted_ss, color='red', label='Mean Prediction')
plt.fill_between(
    Time_Test_Seed_0,
    mean_predicted_ss - 2 * std_predicted_ss,
    mean_predicted_ss + 2 * std_predicted_ss,
    color='red', alpha=0.2, label=r'$\pm$2 Std Dev Uncertainty' # Fixed: raw string for LaTeX-like label
)

plt.xlabel('Time (s)', fontsize=14)
plt.ylabel('Shear Stress (MPa)', fontsize=14)
plt.tick_params(axis='x', labelsize=12)
plt.tick_params(axis='y', labelsize=12)
plt.legend(ncol=3, fontsize=8) # Added ncol=3 and smaller fontsize
plt.axis('tight')

# Save the figure as SVG with 600 DPI
plt.savefig('uncertainty_plot.svg', format='svg', dpi=600)

plt.show()
```
